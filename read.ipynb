```python
fig, axes = plt.subplots(6, 4, figsize=(12, 60))
axes = axes.flatten()  

i = 0
for c in df.select_dtypes(include=[np.number]).columns:
    if c in ['latitude', 'longitude'] or df[c].nunique() <= 2:
        continue
    sns.boxplot(data=df, y=c, ax=axes[i])
    axes[i].set_title(c)
    i += 1

plt.tight_layout()
plt.show()
```

```python
hour_count = df.groupby('hour', as_index= False).agg({'vehicle_count' : 'mean'})


plt.figure(figsize=(10,6))
sns.lineplot(data=hour_count, x='hour', y='vehicle_count', marker='o')
plt.title('Veículos por Hora')
plt.xlabel('Hora do Dia')
plt.ylabel('Total de Veículos')
plt.show()
```
